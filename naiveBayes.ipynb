```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
from yellowbrick.classifier import ConfusionMatrix

# algoritmo de probabilidade
```

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

caminho = '/content/drive/MyDrive/CursoIA/CursoIA/03.Algoritmos_Machine_Learning/insurance.csv'
base = pd.read_csv(caminho, keep_default_na=False)
```

```python
base.head()
base = base.drop(columns=['Unnamed: 0'])
```

```python
base.shape
print(base.isnull().sum()) ##verfica se tem nulos
y = base.iloc[:,7].values #extrai a coluna 8
x = base.drop(base.columns[7], axis=1).values #cria a matrix sem a coluna acima
x
```

```python
labelencoder = LabelEncoder()
for i in range(x.shape[1]):
  if x[:, i].dtype == 'object':
    x[:, i] = labelencoder.fit_transform(x[:, i])

x
```

```python
x_treinamento, x_teste, y_treinamento, y_test = train_test_split(x, y, test_size=0.3, random_state=12) #30% dados para teste
```

```python
modelo = GaussianNB()
modelo.fit(x_treinamento, y_treinamento)
```

```python
previsoes = modelo.predict(x_teste)
```

```python
previsoes

accuracy = accuracy_score(y_test, previsoes)
precision = precision_score(y_test, previsoes, average='weighted')
recall = recall_score(y_test, previsoes, average='weighted')
f1 = f1_score(y_test, previsoes, average='weighted')

print(f"acuracia: {accuracy}, precisao: {precision:0.4f}, recall: {recall}, f1: {f1:0.4f}")
```

```python
report = classification_report(y_test, previsoes)
print(report)
```
