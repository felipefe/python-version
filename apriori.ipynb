```python
#nao supervisionado
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

caminho = '/content/drive/MyDrive/CursoIA/CursoIA/03.Algoritmos_Machine_Learning/Transacoes.txt'

with open(caminho, 'r') as arquivo:
  transacoes = [linha.strip().split(',') for linha in arquivo.readlines()]

transacoes
#
```

```python
# Instanciar o TransactionEncoder
te = TransactionEncoder()
# Ajustar e transformar as transações em uma matriz booleana
te_ary = te.fit(transacoes).transform(transacoes)

# Criar um DataFrame Pandas a partir da matriz
df_transacoes = pd.DataFrame(te_ary, columns=te.columns_)

# Exibir as primeiras linhas do DataFrame preparado
display(df_transacoes.head())
```

```python
frequencia = apriori(df_transacoes, min_support=0.5, use_colnames=True)
display(frequencia)
```

```python
rules = association_rules(frequencia, metric='confidence', min_threshold=0.5)
display(rules)
```
